```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os, json, datetime
```

```python
folder = [path for path in os.listdir('.') if os.path.isdir(os.path.join('.', path))][0]
folder
```

```python
df = pd.read_csv(os.path.join(folder, 'target.csv'))
df
```

```python
plt.figure(figsize=(17, 6))
target = 'ax'
plt.plot(df.time/60, df[target])
plt.xlabel('minutes'); plt.ylabel('values');
plt.grid(.3); plt.title(target); plt.show()

plt.figure(figsize=(17, 6))
target = 'wx'
plt.plot(df.time/60, df[target])
plt.xlabel('minutes'); plt.ylabel('values');
plt.grid(.3); plt.title(target); plt.show()

plt.figure(figsize=(17, 6))
target = 'pitch'
plt.plot(df.time/60, df[target])
plt.xlabel('minutes'); plt.ylabel('values');
plt.grid(.3); plt.title(target); plt.show()

plt.figure(figsize=(17, 6))
target = 'roll'
plt.plot(df.time/60, df[target])
plt.xlabel('minutes'); plt.ylabel('values');
plt.grid(.3); plt.title(target); plt.show()
plt.figure(figsize=(17, 6))
```
