```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import sin, cos, exp, pi
```

```python
DT = 0.1
PERIOD = 5
STOP_TIME = 100
AMPLITUDE = 15

omega = 2*pi/PERIOD
decay = 5/STOP_TIME 

pitch_path = [
    [0 for t in np.arange(0, 20, DT)],
    [AMPLITUDE*sin(omega*t)*exp(-decay*t) for t in np.arange(0, 150, DT)],
    [0 for t in np.arange(0, 50, DT)],
    [AMPLITUDE*sin(omega*t)*exp(-decay*t) for t in np.arange(0, 300, DT)],
]

pitch  = np.concatenate([np.ravel(bloco) for bloco in pitch_path])
tpitch = np.arange(0, len(pitch) * DT, DT)

roll_path = [
    [0 for t in np.arange(0, 130, DT)],
    [AMPLITUDE*sin(omega*t)*exp(-decay*t) for t in np.arange(0, 80, DT)],
    [0 for t in np.arange(0, 100, DT)],
    [AMPLITUDE*sin(omega*t)*exp(-decay*t) for t in np.arange(0, 80, DT)],
]

roll  = np.concatenate([np.ravel(bloco) for bloco in roll_path])
troll = np.arange(0, len(roll) * DT, DT)

plt.figure(figsize=(20, 6))
plt.plot(tpitch, pitch, label='pitch')
plt.plot(troll,  roll,  label='roll')
plt.grid(alpha=.3); plt.legend()
plt.show()
```
